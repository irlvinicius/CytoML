# Calibração do CLAHE

Gera comparações antes/depois (mediana + CLAHE) em amostras variadas e em vários valores de `clahe_clip`, para escolher o parâmetro por inspeção visual sem estourar o contraste.

```python
import sys
from pathlib import Path

REPO_ROOT = Path("../../..").resolve()
sys.path.insert(0, str(REPO_ROOT / "src" / "preprocessing"))

from filters import save_comparison

RAW_ROOT = REPO_ROOT / "data" / "raw" / "24_chromosomes_object"
FIGURES_OUT = REPO_ROOT / "outputs" / "figures" / "clahe_calibration"
FIGURES_OUT.mkdir(parents=True, exist_ok=True)

# Amostras variadas (algumas do diff_image.txt, marcadas como "difíceis" pelos
# autores do dataset — fundo escuro / overlap, exatamente o cenário problemático).
diff_names = [n.strip() for n in (RAW_ROOT.parent / "diff_image.txt").read_text().splitlines() if n.strip()]
regular_names = [p.stem for p in list((RAW_ROOT / "JEPG").glob("*.jpg"))[:3]]
sample_names = regular_names + diff_names[:3]
print("Amostras selecionadas:", sample_names)
```

Gera o grid antes/depois pra cada amostra e cada `clahe_clip`, salvando em `outputs/figures/clahe_calibration/` (`.gitignore`).

```python
CLAHE_CLIP_VALUES = [1.0, 1.5, 2.0]

for name in sample_names:
    image_path = RAW_ROOT / "JEPG" / f"{name}.jpg"
    if not image_path.exists():
        print(f"pulando {name}: imagem não encontrada no mirror local")
        continue
    for clip in CLAHE_CLIP_VALUES:
        out_path = FIGURES_OUT / f"{name}_clip{clip}.png"
        save_comparison(image_path, out_path, clahe_clip=clip)
        print(f"salvo: {out_path}")
```
